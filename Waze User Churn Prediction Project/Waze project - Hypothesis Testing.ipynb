### **Data Exploration and Hypothesis Testing**

**Goal:**
To apply descriptive statistics and hypothesis testing using Python.




**Research Question:**
*Do drivers who use the application on an iPhone have the same average number of rides as drivers who use Android devices?*



```python
# Import any relevant packages or libraries
import pandas as pd
from scipy import stats
```

```python
# Load dataset into dataframe
df = pd.read_csv("waze_dataset.csv")
```

```python
# Create `map_dictionary`
map_dictionary = {'Android': 2, 'iPhone': 1}

# Create new `device_type` column
df['device_type'] = df['device']

# Map the new column to the dictionary
df['device_type'] = df['device_type'].map(map_dictionary)

df['device_type'].head()
```

```python
df.groupby('device_type')['drives'].mean()
```

Based on the averages, drivers who use an iPhone appear to have a higher average number of rides compared to Android users. However, this observed difference may be due to random variation in the sample rather than a true difference in behavior. To determine if the difference is statistically significant, a hypothesis test should be conducted.




### **Hypothesis Testing**

The goal is to conduct a two-sample t-test. The steps for a hypothesis test are:

1. State the null and alternative hypotheses
2. Choose a significance level
3. Calculate the p-value
4. Decide whether to reject or fail to reject the null hypothesis

**Note:** A two-sample t-test is appropriate here because the two groups (iPhone users and Android users) are independent.

#### Hypotheses

* \$H\_0\$: There is no difference in the average number of drives between iPhone and Android users.
* \$H\_A\$: There is a difference in the average number of drives between iPhone and Android users.

**5% significance level** 




```python
# Isolate the `drives` column for iPhone users.
iPhone = df[df['device_type'] == 1]['drives']

# Isolate the `drives` column for Android users.
Android = df[df['device_type'] == 2]['drives']

# Perform the t-test
stats.ttest_ind(a=iPhone, b=Android, equal_var=False)
```

Based on the p-value you obtained, do you reject or fail to reject the null hypothesis?

Since the p-value is greater than the 5% significance level, you fail to reject the null hypothesis. This means there is no statistically significant difference in the average number of drives between iPhone users and Android users.
The analysis shows that iPhone and Android users take a similar number of drives on average. There is no statistically significant difference between the two groups.
