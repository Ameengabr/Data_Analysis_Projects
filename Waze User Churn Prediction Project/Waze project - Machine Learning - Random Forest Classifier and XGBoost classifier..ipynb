```python
def make_results(model_name:str, model_object, metric:str):


    # Create dictionary that maps input metric to actual metric name in GridSearchCV
    metric_dict = {'precision': 'mean_test_precision',
                   'recall': 'mean_test_recall',
                   'f1': 'mean_test_f1',
                   'accuracy': 'mean_test_accuracy',
                   }

    # Get all the results from the CV and put them in a df
    cv_results = pd.DataFrame(model_object.cv_results_)

    # Isolate the row of the df with the max(metric) score
    best_estimator_results = cv_results.iloc[cv_results[metric_dict[metric]].idxmax(), :]

    # Extract accuracy, precision, recall, and f1 score from that row
    f1 = best_estimator_results.mean_test_f1
    recall = best_estimator_results.mean_test_recall
    precision = best_estimator_results.mean_test_precision
    accuracy = best_estimator_results.mean_test_accuracy

    # Create table of results
    table = pd.DataFrame({'model': [model_name],
                          'precision': [precision],
                          'recall': [recall],
                          'F1': [f1],
                          'accuracy': [accuracy],
                          },
                         )

    return table
```

```python
results = make_results('RF cv', rf_cv, 'recall')
results
```

```python
def get_test_scores(model_name:str, preds, y_test_data):

    accuracy = accuracy_score(y_test_data, preds)
    precision = precision_score(y_test_data, preds)
    recall = recall_score(y_test_data, preds)
    f1 = f1_score(y_test_data, preds)

    table = pd.DataFrame({'model': [model_name],
                          'precision': [precision],
                          'recall': [recall],
                          'F1': [f1],
                          'accuracy': [accuracy]
                          })

    return table
```

#### **Identify an Optimal Decision Threshold**

Most classification algorithms, including those in scikit-learn, use a default decision threshold of 0.5. For the Waze models, this means that a user is predicted to churn (assigned a value of `1`) if the model estimates a 50% or higher probability of churn.

However, when working with imbalanced datasets where the target class is a minority, this default threshold may not be optimal. A **precision-recall curve** can be used to visualize the trade-off between precision and recall and help identify a more suitable threshold for the business objective.

The precision-recall curve for the XGBoost champion model on the test data illustrates how adjusting the threshold affects the model’s performance in identifying users likely to churn.


```python
# Plot precision-recall curve
display = PrecisionRecallDisplay.from_estimator(
    xgb_cv.best_estimator_, X_test, y_test, name='XGBoost'
    )
plt.title('Precision-recall curve, XGBoost model');
```

When recall increases, precision typically decreases. However, if false positives carry minimal risk, adjusting the decision threshold can be beneficial.

In the context of the Waze project, a false positive simply means a user who would not churn receives an email or banner notification—this poses very low risk.

By lowering the default decision threshold from 0.5 to, for example, 0.4, the model can identify more potential churners, increasing recall while accepting a slight decrease in precision.

Using the precision-recall curve as a guide, suppose the goal is to achieve a **recall of 0.5** and accept a corresponding **precision of approximately 30%**. This means the model would correctly identify half of the users who will actually churn, even though only about 30% of users predicted to churn are truly at risk.

To determine the threshold that yields this balance, you can use various approaches, including a function that calculates the threshold based on the desired recall level.


```python
def threshold_finder(y_test_data, probabilities, desired_recall):
    probs = [x[1] for x in probabilities]  # Isolate second column of `probabilities`
    thresholds = np.arange(0, 1, 0.001)    # Set a grid of 1,000 thresholds to test

    scores = []
    for threshold in thresholds:
        # Create a new array of {0, 1} predictions based on new threshold
        preds = np.array([1 if x >= threshold else 0 for x in probs])
        # Calculate recall score for that threshold
        recall = recall_score(y_test_data, preds)
        # Append the threshold and its corresponding recall score as a tuple to `scores`
        scores.append((threshold, recall))

    distances = []
    for idx, score in enumerate(scores):
        # Calculate how close each actual score is to the desired score
        distance = abs(score[1] - desired_recall)
        # Append the (index#, distance) tuple to `distances`
        distances.append((idx, distance))

    # Sort `distances` by the second value in each of its tuples (least to greatest)
    sorted_distances = sorted(distances, key=lambda x: x[1], reverse=False)
    # Identify the tuple with the actual recall closest to desired recall
    best = sorted_distances[0]
    # Isolate the index of the threshold with the closest recall score
    best_idx = best[0]
    # Retrieve the threshold and actual recall score closest to desired recall
    threshold, recall = scores[best_idx]

    return threshold, recall
```
